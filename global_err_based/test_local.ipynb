```python
import numpy as np
import matplotlib.pyplot as plt
import scipy as sp
import newton_local_deuflhard as local_d
import newton_local_classical as local
```

```python
# Base test case: A simple nonlinear system
def func(x):
    return np.array([
        x[0]**2 + x[1]**2 - 4,
        x[1] - x[0]**2 + 1
    ])

def jac(x):
    return np.array([
        [2*x[0], 2*x[1]],
        [-2*x[0], 1]
    ])



x0 = np.array([0.5, 0.5])

success_d, x_d = local_d.solve(
    func, jac, x0, XTOL = 1e-6, MAX_ITER = 20
)

success_c, x_c = local.solve(
    func, jac, x0, RTOL = 1e-6, MAX_ITER = 20
)


print("Deuflhard's method:")
print(f"Success: {success_d}, Solution: {x_d}")

print("Classical Newton's method:")
print(f"Success: {success_c}, Solution: {x_c}")
```

```python
# Test a badly scaled problem
A = 1e12

def func(x):
    return np.array([
        A * (x[0] - 1.0)**2
    ])

def jac(x):
    return np.array([
        [2.0 * A * (x[0] - 1.0)]
    ])

x0 = np.array([1.1])

success_d, x_d = local_d.solve(
    func, jac, x0, XTOL=1e-6, MAX_ITER=20
)

success_c, x_c = local.solve(
    func, jac, x0, RTOL=1e-6, MAX_ITER=20
)
print("Deuflhard's method:")
print(f"Success: {success_d}, Solution: {x_d}, Residual: {np.linalg.norm(func(x_d))}")

print("Classical Newton's method:")
print(f"Success: {success_c}, Solution: {x_c}, Residual: {np.linalg.norm(func(x_c))}")
```
